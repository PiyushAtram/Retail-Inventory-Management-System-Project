```python
%sql
create or replace table  fisrt_cat.gold.product_summery as
select p.*,i.Stock,i.Minimum_Order_Quantity,i.Status from fisrt_cat.silver.inventory i join fisrt_cat.silver.products p on p.Product_Id=i.Product_Id
```

```python
%sql
select * from fisrt_cat.gold.product_summery
```

```python
%sql
select Status , count(*) from fisrt_cat.gold.product_summery group by Status order by count(*) desc
```

```python
%sql
select Category, avg(Price) from fisrt_cat.gold.product_summery GROUP BY Category
```

```python
%sql
select count(*) from fisrt_cat.gold.product_summery
```

```python
%sql
select Brand,sum(Price) from fisrt_cat.gold.product_summery group by Brand order by sum(Price) desc
```

```python
df = spark.sql("""
SELECT 
    Product_Id,
    Product_Name,
    Stock,
    Status
FROM fisrt_cat.gold.product_summery
WHERE Status IN ('LOW STOCK','OVERSTOCK')
""")

result = df.toJSON().collect()

dbutils.notebook.exit(str(result))
```
